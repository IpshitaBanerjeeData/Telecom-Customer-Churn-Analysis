```python
import pandas as pd
import numpy as np
```

```python
# 1. Load the raw dataset
# (Make sure your CSV file is in the same directory or provide the correct path)
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
print(f"Original dataset shape: {df.shape}")
```

```python
# 2. Handle the 'TotalCharges' quirk
# This column is stored as an object/string because brand-new customers (tenure = 0)
# have empty spaces (" ") instead of numbers.
# 'errors="coerce"' forces these spaces into NaN (Not a Number).
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
# Inspect how many null values were generated
print("Null values in TotalCharges:", df['TotalCharges'].isnull().sum())
```

```python
# 3. Handle missing values
# Since these rows correspond to customers with tenure = 0, we fill their TotalCharges with 0.
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
# Double-check if there are any other missing values in the rest of the dataset
print("\nMissing values across all columns:\n", df.isnull().sum())
```

```python
# 4. Drop unnecessary columns
# customerID is a unique identifier and won't help in pattern analysis or modeling.
if 'customerID' in df.columns:
    df = df.drop(columns=['customerID'])
```

```python
# 5. Create a numerical target column for easier aggregations later
df['Churn_Numeric'] = df['Churn'].map({'Yes': 1, 'No': 0})
```

```python
print(f"\nCleaned dataset shape: {df.shape}")
```

```python
# 6. Save the clean data
df.to_csv("telco_churn_cleaned.csv", index=False)
print("\nSuccess! Cleaned dataset saved as 'telco_churn_cleaned.csv'.")
```
